## Prueba de cálculo económico

Se calcula el factor de capacidad con los datos del ejercicio inicial.

```python
capacidad_mw = 363
energia_anual_mwh = 972_915
horas_anuales = 8_760

factor_capacidad = energia_anual_mwh / (capacidad_mw * horas_anuales)
assert np.isclose(factor_capacidad, 0.30596, atol=1e-5)
print(f'Factor de capacidad: {factor_capacidad:.2%}')
```

## Prueba de las herramientas solares

Se obtiene la posición solar para una ubicación y una fecha conocidas. La prueba no necesita conexión a internet.

```python
ubicacion = pvlib.location.Location(19.054, -98.283, tz='America/Mexico_City', altitude=2_130)
fecha = pd.DatetimeIndex(['2026-03-21 12:00'], tz=ubicacion.tz)
posicion_solar = ubicacion.get_solarposition(fecha)

display(posicion_solar[['apparent_zenith', 'azimuth']])
assert posicion_solar[['apparent_zenith', 'azimuth']].notna().all().all()
print('Entorno listo para el curso.')
```
